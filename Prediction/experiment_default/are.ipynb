```python
import pandas as pd

# Texte du tableau
table_text = ["""
FWI & 0.15, 0.14 & 0.84 & 0.09 & 0.05, 0.05 & - \\
LR & 0.42, 0.14 & 0.50 & 0.36 & 0.24, 0.08 & 0.23, 0.08 \\
XGBoost & 0.42, 0.14 & 0.52 & 0.36 & 0.24, 0.08 & 0.24, 0.07 \\
CatBoost & 0.43, 0.13 & 0.52 & 0.37 & 0.24, 0.07 & 0.23, 0.06 \\
MLP & 0.42, 0.14 & 0.60 & 0.32 & 0.24, 0.08 & 0.23, 0.06 \\
GRUGNN & 0.41, 0.14 & 0.66 & 0.24 & 0.24, 0.08 & 0.24, 0.08 \\
GraphCast & 0.41, 0.16 & 0.61 & 0.31 & 0.23, 0.09 & 0.22, 0.07 \\
LSTM & 0.41, 0.15 & 0.64 & 0.30 & 0.23, 0.09 & 0.23, 0.07 \\
GRU & 0.43, 0.15 & 0.57 & 0.35 & 0.24, 0.09 & 0.23, 0.07 \\
DilatedCNN & 0.40, 0.16 & 0.54 & 0.32 & 0.22, 0.09 & 0.22, 0.07 \\
CatBoost Binary & 0.42, 0.15 & 0.63 & 0.31 & - & - \\
MLP Binary & 0.40, 0.15 & 0.58 & 0.32 & - & - \\
GRU Binary & 0.37, 0.17 & 0.76 & 0.24 & - & - \\
""",
"""FWI & 0.38, 0.38 & 0.84 & 0.29 & 0.17, 0.17 & - \\
LR & 0.63, 0.33 & 0.68 & 0.59 & 0.39, 0.20 & 0.41, 0.20 \\
XGBoost & 0.63, 0.34 & 0.68 & 0.59 & 0.39, 0.20 & 0.42, 0.20 \\
CatBoost & 0.64, 0.35 & 0.69 & 0.59 & 0.40, 0.21 & 0.42, 0.20 \\
MLP & 0.61, 0.37 & 0.73 & 0.52 & 0.34, 0.21 & 0.41, 0.19 \\
GRUGNN & 0.59, 0.37 & 0.80 & 0.47 & 0.34, 0.21 & 0.40, 0.21 \\
GraphCast & 0.61, 0.34 & 0.66 & 0.56 & 0.37, 0.20 & 0.40, 0.20 \\
LSTM & 0.58, 0.36 & 0.72 & 0.47 & 0.37, 0.22 & 0.40, 0.19 \\
GRU & 0.60, 0.37 & 0.75 & 0.50 & 0.37, 0.22 & 0.41, 0.21 \\
DilatedCNN & 0.60, 0.37 & 0.71 & 0.52 & 0.38, 0.22 & 0.39, 0.20 \\
CatBoost Binary & 0.60, 0.40 & 0.69 & 0.59 & - & - \\
MLP Binary & 0.54, 0.39 & 0.86 & 0.40 & - & - \\
GRU Binary & 0.58, 0.38 & 0.75 & 0.48 & - & - \\
    """,
    """FWI & 0.47, 0.47 & 0.81 & 0.39 & 0.22, 0.22 & - \\
LR & 0.67, 0.44 & 0.72 & 0.62 & 0.41, 0.27 & 0.44, 0.27 \\
XGBoost & 0.67, 0.42 & 0.71 & 0.64 & 0.41, 0.25 & 0.44, 0.26 \\
CatBoost & 0.68, 0.43 & 0.73 & 0.67 & 0.43, 0.26 & 0.45, 0.28 \\
MLP & 0.63, 0.47 & 0.82 & 0.51 & 0.40, 0.29 & 0.45, 0.25 \\
GRUGNN & 0.64, 0.45 & 0.80 & 0.53 & 0.40, 0.27 & 0.42, 0.28 \\
GraphCast & 0.63, 0.46 & 0.80 & 0.52 & 0.39, 0.28 & 0.43, 0.29 \\
LSTM & 0.63, 0.44 & 0.76 & 0.54 & 0.40, 0.27 & 0.43, 0.26 \\
GRU & 0.64, 0.45 & 0.76 & 0.56 & 0.40, 0.27 & 0.44, 0.27 \\
DilatedCNN & 0.64, 0.44 & 0.72 & 0.57 & 0.40, 0.26 & 0.42, 0.26 \\
CatBoost Binary & 0.64, 0.50 & 0.86 & 0.51 & - & - \\
MLP Binary & 0.63, 0.46 & 0.78 & 0.53 & - & - \\
GRU Binary & 0.61, 0.46 & 0.79 & 0.50 & - & - \\""",
"""FWI & 0.47, 0.55 & 0.81 & 0.47 & 0.25, 0.25 & - \\
LR & 0.69, 0.54 & 0.77 & 0.64 & 0.43, 0.33 & 0.46, 0.35 \\
XGBoost & 0.71, 0.50 & 0.71 & 0.71 & 0.44, 0.31 & 0.46, 0.33 \\
CatBoost & 0.71, 0.53 & 0.76 & 0.67 & 0.45, 0.34 & 0.48, 0.35 \\
MLP & 0.69, 0.53 & 0.77 & 0.63 & 0.42, 0.33 & 0.46, 0.34 \\
GRUGNN & 0.68, 0.53 & 0.81 & 0.59 & 0.43, 0.33 & 0.45, 0.35 \\
GraphCast & 0.67, 0.47 & 0.68 & 0.67 & 0.41, 0.29 & 0.45, 0.35 \\
LSTM & 0.68, 0.54 & 0.77 & 0.61 & 0.43, 0.34 & 0.44, 0.35 \\
GRU & 0.67, 0.52 & 0.76 & 0.61 & 0.43, 0.33 & 0.44, 0.34 \\
DilatedCNN & 0.66, 0.53 & 0.72 & 0.61 & 0.41, 0.33 & 0.41, 0.34 \\
CatBoost Binary & 0.65, 0.58 & 0.91 & 0.51 & - & - \\
MLP Binary & 0.66, 0.57 & 0.84 & 0.55 & - & - \\
GRU Binary & 0.67, 0.57 & 0.82 & 0.57 & - & - \\"""
]
# Noms de colonnes
columns = [
    "Model",
    "IoU FO (Sample, Dept)",
    "Metric2",
    "Metric3",
    "Metric4 (Sample, Dept)",
    "Metric5 (Sample, Dept)"
]

# Colonnes à parser comme tuple et à calculer l'aire
import pandas as pd

def parse_model_table(table_text, tuple_columns=None):
    """
    Parse un tableau de résultats de modèles et calcule l'aire entre les valeurs des colonnes tuple.

    Args:
        table_text (str): Texte brut du tableau avec & comme séparateur.
        tuple_columns (list): Liste des noms de colonnes à parser comme (x, y). Ex: ['IoU FO', 'IoU BA']

    Returns:
        pd.DataFrame: DataFrame structuré avec colonnes supplémentaires pour l'aire.
    """
    # Colonnes
    columns = ["Model", "F1", "rec", "prec", "IoU FO", "IoU BA"]

    # Par défaut, parser les colonnes 'IoU FO' et 'IoU BA' comme tuples
    if tuple_columns is None:
        tuple_columns = ["F1", "IoU FO", "IoU BA"]

    # Nettoyage
    cleaned_text = table_text.replace('\\', '').strip()
    rows = []
    for line in cleaned_text.split('\n'):
        parts = [p.strip() for p in line.split('&')]
        model = parts[0]
        values = []
        for i, p in enumerate(parts[1:]):
            col_name = columns[i + 1]
            if col_name in tuple_columns and ',' in p:
                try:
                    nums = [float(x.strip()) for x in p.split(',')]
                    values.append(tuple(nums))
                except:
                    values.append(None)
            elif p == '-' or p == '':
                values.append(None)
            else:
                try:
                    values.append(float(p))
                except ValueError:
                    values.append(None)
        rows.append([model] + values)

    # DataFrame
    df = pd.DataFrame(rows, columns=columns)

    # Calcul des aires
    for col in tuple_columns:
        df[f"{col} - Samples By Departments"] = df[col].apply(lambda x: round(x[0] * x[1], 3) if isinstance(x, tuple) else None)

    return df


test_name = [0,7,15,31]
df = []
for i, name in enumerate(test_name):
    dff = parse_model_table(table_text[i])
    dff['test'] = name
    df.append(dff)

df = pd.concat(df).reset_index()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D
import math

def plot_multiple_metrics_subplots(df, columns, name_col='Model', hue_col='test', figsize=(6, 5)):
    """
    Crée une figure avec 1 subplot par métrique (disposés sur 2 lignes si nécessaire),
    sans cases vides. Chaque subplot affiche les barres groupées par modèle et test,
    avec croix rouge pour le max et bleue pour le 2e max.

    Args:
        df (pd.DataFrame): Données.
        columns (list): Liste des colonnes de métriques à tracer.
        name_col (str): Colonne des modèles.
        hue_col (str): Colonne de regroupement (ex: test).
        figsize (tuple): Taille d'un seul subplot (largeur, hauteur).
    """
    num_metrics = len(columns)
    n_cols = math.ceil(num_metrics / 2)
    n_rows = math.ceil(num_metrics / n_cols)

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(figsize[0] * n_cols, figsize[1] * n_rows), sharey=False)
    axes = axes.flatten() if num_metrics > 1 else [axes]

    for idx, column in enumerate(columns):
        ax = axes[idx]
        data = df[[name_col, hue_col, column]].dropna()
        sns.barplot(data=data, x=name_col, y=column, hue=hue_col, ax=ax, ci=None)

        x_labels = [t.get_text() for t in ax.get_xticklabels()]
        hue_order = sorted(data[hue_col].unique())
        total_hue = len(hue_order)
        width = 0.8 / total_hue

        for group_val, group_df in data.groupby(hue_col):
            sorted_group = group_df.sort_values(by=column, ascending=False)
            for rank, (color, marker) in zip([0, 1], [('red', 'x'), ('blue', 'x')]):
                if len(sorted_group) > rank:
                    row = sorted_group.iloc[rank]
                    model = row[name_col]
                    value = row[column]
                    x_index = x_labels.index(model)
                    hue_shift = hue_order.index(group_val)
                    x_coord = x_index - 0.4 + width / 2 + hue_shift * width
                    ax.plot(x_coord, value, marker=marker, color=color, markersize=10, markeredgewidth=2)

        ax.set_title(column, fontsize=15)
        ax.set_xlabel(name_col, fontsize=13)
        ax.set_ylabel("Value", fontsize=13)
        ax.tick_params(axis='both', labelsize=12)
        ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
        ax.get_legend().remove()

    # Supprimer les axes restants inutilisés
    for j in range(num_metrics, len(axes)):
        fig.delaxes(axes[j])

    # Légende globale unique
    last_ax = axes[min(num_metrics - 1, len(axes) - 1)]
    handles_bar, labels_bar = last_ax.get_legend_handles_labels()
    custom_handles = [
        Line2D([0], [0], color='red', marker='x', linestyle='None', markersize=10, label='Max'),
        Line2D([0], [0], color='blue', marker='x', linestyle='None', markersize=10, label='2nd Max')
    ]
    handles_all = handles_bar + custom_handles
    labels_all = labels_bar + ['Max', '2nd Max']

    fig.legend(handles_all, labels_all, loc='upper center', ncol=len(labels_all), fontsize=12, title="Legend", title_fontsize=13)
    plt.tight_layout(rect=[0, 0, 1, 0.92])
    plt.show()
```

```python
plot_multiple_metrics_subplots(df[df['Model'].isin(['GRU', 'GRUGNN'])], ['IoU FO - Samples By Departments', 'F1 - Samples By Departments', 'IoU BA - Samples By Departments'])
```
