```python
import os
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

base_path = '/media/caron/X9 Pro/travaille/Thèse/Model/HexagonalScale/ST-GNN-for-wildifre-prediction/Prediction/Target/firepoint'

dept_mapping = {
    ('BDIFF', 'Bouches-du-Rhône'): '13-bouches-du-rhone',
    ('BDIFF', 'Hérault'): '34-herault',
    ('Firefighter', 'Ain'): '01-ain',
    ('Firefighter', 'Doubs'): '25-doubs',
    ('Firefighter', 'Yvelines'): '78-yvelines',
}

stats = []

for idx, row in df_params[['Database', 'Department']].drop_duplicates().iterrows():
    db = row['Database']
    dept = row['Department']
    slug = dept_mapping.get((db, dept))
    if not slug:
        continue
        
    db_folder = "bdiff" if db == "BDIFF" else "firemen"
    path = os.path.join(base_path, db_folder, "datacube", f"departement-{slug}", "2x2", "datacube.pkl")
    
    if os.path.exists(path):
        print(f"Loading {path}...")
        with open(path, 'rb') as f:
            log = pickle.load(f)
            
        # Sum over date dimension
        if 'date' in log.dims:
            occ_sum = log['occurence'].sum(dim='date')
        elif 'time' in log.dims:
            occ_sum = log['occurence'].sum(dim='time')
        else:
            occ_sum = log['occurence'].sum() 
            
        arr = occ_sum.values
        arr = arr[~np.isnan(arr)] 
        
        stat_mean = arr.mean()
        stat_max = arr.max()
        stat_range = arr.max() - arr.min()
        area_pixels = len(arr)
        positive_pixels = (arr > 0).sum()

        y_pos, x_pos = np.where(occ_sum.values > 0)
        if len(y_pos) > 1:
            from scipy.spatial.distance import pdist
            pos_coords = np.column_stack((y_pos, x_pos))
            avg_pos_dist = pdist(pos_coords).mean()
        else:
            avg_pos_dist = 0.0
                
        stats.append({
            'Database': db,
            'Department': dept,
            'occurence_mean': stat_mean,
            'occurence_max': stat_max,
            'occurence_range': stat_range,
            'area_pixels': area_pixels,
            'positive_pixels': positive_pixels,
            'avg_pos_dist': avg_pos_dist
        })

df_stats = pd.DataFrame(stats)
display(df_stats)

df_merged = pd.merge(df_params, df_stats, on=['Database', 'Department'], how='inner')

x_metrics = ['occurence_mean', 'occurence_max', 'area_pixels', 'positive_pixels', 'avg_pos_dist']
y_params = ['nb_attempt', 'n_reduce_class']

for threshold in [0.2, 0.3, 0.4, 'All']:
    if threshold == 'All':
        df_thresh = df_merged
        title = "All Thresholds Combined"
    else:
        if 'Threshold' in df_merged.columns:
            df_thresh = df_merged[df_merged['Threshold'] == threshold]
            title = f"Threshold = {threshold}"
        else:
            df_thresh = df_merged
            title = f"Threshold = {threshold}"
            
    fig, axes = plt.subplots(len(y_params), len(x_metrics), figsize=(5 * len(x_metrics), 5 * len(y_params)))
    fig.suptitle(title, fontsize=16)

    for i, y_col in enumerate(y_params):
        for j, x_col in enumerate(x_metrics):
            ax = axes[i, j]
            df_plot = df_thresh.groupby([x_col, y_col, 'Department']).size().reset_index(name='count')
            sns.scatterplot(data=df_plot, x=x_col, y=y_col, hue='Department', size='count', sizes=(50, 500), ax=ax, alpha=0.7, legend=(i==0 and j==len(x_metrics)-1))
            if i==0 and j==len(x_metrics)-1:
                ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
            ax.set_title(f"{y_col} vs {x_col}")
            
    plt.tight_layout()
    plt.subplots_adjust(top=0.95)
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import spearmanr

print("Starting Advanced Statistical Analysis Pipeline...")

# 1. Base Data & Derived Metrics (Form vs. Size)
df_ana = df_merged.copy()



# Calculate A and R from n_reduce_class since they might not be in df_merged


# A and R are short names for nb_attempt and n_reduce_class
if 'A' not in df_ana.columns:
    df_ana['A'] = df_ana['nb_attempt']
if 'R' not in df_ana.columns:
    df_ana['R'] = df_ana['n_reduce_class']


    


# Ensure we don't divide by zero
epsilon = 1e-6

df_ana['rho'] = df_ana['positive_pixels'] / (df_ana['area_pixels'] + epsilon)
df_ana['kappa'] = df_ana['occurence_max'] / (df_ana['occurence_mean'] + epsilon)
df_ana['delta'] = df_ana['positive_pixels'] / (df_ana['avg_pos_dist'] + epsilon)

print("Calculated derived metrics: rho, kappa, delta")

# 2. Stability Metrics (Threshold)
agg_funcs = {
    'area_pixels': ['std', lambda x: np.max(x) - np.min(x)],
    'positive_pixels': ['std', lambda x: np.max(x) - np.min(x)],
    'avg_pos_dist': ['std', lambda x: np.max(x) - np.min(x)],
    'rho': ['std', lambda x: np.max(x) - np.min(x)]
}

df_stability = df_ana.groupby(['Database', 'Department', 'Model', 'A', 'R']).agg(agg_funcs).reset_index()
# Flatten MultiIndex columns
df_stability.columns = ['_'.join(col).strip('_') if type(col) is tuple else col for col in df_stability.columns.values]
df_stability.rename(columns=lambda x: x.replace('<lambda_0>', 'range'), inplace=True)

print("Calculated stability metrics across thresholds")

# 3. Department Standardization (Removing Base Effect)
cols_to_norm = ['occurence_mean', 'occurence_max', 'area_pixels', 'positive_pixels', 'avg_pos_dist', 'rho', 'kappa', 'delta']
for col in cols_to_norm:
    mean_dept = df_ana.groupby(['Database', 'Department'])[col].transform('mean')
    std_dept = df_ana.groupby(['Database', 'Department'])[col].transform('std')
    df_ana[col + '_z'] = (df_ana[col] - mean_dept) / (std_dept + epsilon)

print("Standardized metrics by department")

# 4. Exploratory Visualizations
def plot_analysis():
    # 4.1 Normalized Scatterplots (A and R vs derived variables)
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    fig.suptitle('Normalized Variables vs A and R')
    
    var_list = ['rho', 'kappa', 'delta']
    for i, param in enumerate(['A', 'R']):
        for j, var in enumerate(var_list):
            sns.scatterplot(data=df_ana, x=var+'_z', y=param, hue='Department', ax=axes[i, j], legend=False)
            axes[i, j].set_title(f"{param} vs Normalized {var}")
            
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

    # 4.2 Heatmap of A, R couples by Department
    plt.figure(figsize=(10, 6))
    heatmap_data = df_ana.groupby(['A', 'R']).size().reset_index(name='count')
    # Make sure we don't hit pivoting errors if dimensions change, use pivot_table with sum
    heatmap_pivot = heatmap_data.pivot_table(index='A', columns='R', values='count', aggfunc='sum').fillna(0)
    sns.heatmap(heatmap_pivot, cmap='viridis', annot=True, fmt='g')
    plt.title('Global Frequency of (A, R) configurations')
    plt.show()
    
    # 4.3 Distribution conditional to A and R
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    fig.suptitle('Distribution of internal structure vars vs A and R')
    for i, param in enumerate(['A', 'R']):
        for j, var in enumerate(var_list):
            sns.boxplot(data=df_ana, x=param, y=var, ax=axes[i, j])
            axes[i, j].set_title(f"{var} by {param}")
            
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

plot_analysis()

# 5. Non-Linear Correlation Testing
print("\n--- Spearman Rank Correlations with A and R ---\n")
for var in cols_to_norm + [c+'_z' for c in cols_to_norm]:
    corr_a, p_a = spearmanr(df_ana['A'], df_ana[var])
    corr_r, p_r = spearmanr(df_ana['R'], df_ana[var])
    print(f"{var:25s} -> A: corr={corr_a:+.3f} (p={p_a:.3f}) | R: corr={corr_r:+.3f} (p={p_r:.3f})")

print("\n--- Stability Metrics Correlations ---\n")
for var in df_stability.columns:
    if var not in ['Database', 'Department', 'Model', 'A', 'R']:
        corr_a, p_a = spearmanr(df_stability['A'], df_stability[var])
        corr_r, p_r = spearmanr(df_stability['R'], df_stability[var])
        print(f"{var:25s} -> A: corr={corr_a:+.3f} (p={p_a:.3f}) | R: corr={corr_r:+.3f} (p={p_r:.3f})")
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree

print("\n--- Decision Tree Analysis (Interpretable Rules) ---\n")

# Prepare data
features = ['rho', 'kappa', 'delta']
X_tree = df_ana[features].fillna(0) # Handle potential NaNs

# Try to predict R 
y_tree_R = df_ana['R']

if len(y_tree_R.unique()) > 1:
    clf_R = DecisionTreeClassifier(max_depth=10, random_state=42, class_weight='balanced')
    clf_R.fit(X_tree, y_tree_R)

    plt.figure(figsize=(20, 10))
    plot_tree(clf_R, feature_names=features, class_names=[str(c) for c in clf_R.classes_], filled=True, rounded=True, proportion=False)
    plt.title("Decision Tree: Predicting R based on structural metrics (rho, kappa, delta)", fontsize=16)
    plt.show()

    print("Feature importances for R prediction:")
    for feat, imp in zip(features, clf_R.feature_importances_):
        print(f"  {feat}: {imp:.3f}")
else:
    print("Not enough variance in R to train a decision tree.")

# Try to predict A
y_tree_A = df_ana['A']
if len(y_tree_A.unique()) > 1:
    clf_A = DecisionTreeClassifier(max_depth=3, random_state=42, class_weight='balanced')
    clf_A.fit(X_tree, y_tree_A)

    plt.figure(figsize=(20, 10))
    plot_tree(clf_A, feature_names=features, class_names=[str(c) for c in clf_A.classes_], filled=True, rounded=True, proportion=False)
    plt.title("Decision Tree: Predicting A based on structural metrics (rho, kappa, delta)", fontsize=16)
    plt.show()
    
    print("\nFeature importances for A prediction:")
    for feat, imp in zip(features, clf_A.feature_importances_):
        print(f"  {feat}: {imp:.3f}")
else:
    print("Not enough variance in A to train a decision tree.")
```
